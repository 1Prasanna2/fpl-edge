```python
import json
import pandas as pd
from pathlib import Path

def find_project_root():
    here = Path.cwd().parent
    for folder in [here, *here.parents]:
        if (folder / "requirements.txt").exists() or (folder / ".git").exists():
            return folder
        return here

PROJECT_ROOT = find_project_root()
print(PROJECT_ROOT)
BOOTSTRAP_FILEPATH = PROJECT_ROOT / "data" / "raw" / "bootstrap-static.json"
with open(BOOTSTRAP_FILEPATH,"r",encoding="utf-8") as f:
    bootstrap = json.load(f)
    
FIXTURES_FILEPATH = PROJECT_ROOT / "data" / "raw" / "fixtures.json"
with open(FIXTURES_FILEPATH,"r",encoding="utf-8") as f:
    fixtures = json.load(f)
    
players = pd.json_normalize(bootstrap["elements"])
teams = pd.json_normalize(bootstrap["teams"])
events = pd.json_normalize(bootstrap["events"])
fixtures_df = pd.json_normalize(fixtures)

players.head()
```
